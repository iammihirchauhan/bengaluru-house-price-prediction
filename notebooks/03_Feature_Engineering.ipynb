```python
import pandas as pd

df = pd.read_csv("../data/bengaluru_model_data.csv")

print("Shape:", df.shape)
print(df["location"].nunique())
```

```python
location_counts = df["location"].value_counts()

print("Total unique locations:", len(location_counts))

print("\nTop 20 locations:")
print(location_counts.head(20))

print("\nLocations with 10 or fewer properties:")
print(location_counts[location_counts <= 10].shape)
```

```python
location_counts = df["location"].value_counts()

rare_locations = location_counts[location_counts <= 10].index

df["location"] = df["location"].apply(
    lambda x: "Other" if x in rare_locations else x
)

print("Unique locations after grouping:", df["location"].nunique())
print("\nLocation distribution:")
print(df["location"].value_counts().head(20))
```

```python
df.to_csv("../data/bengaluru_featured.csv", index=False)

print("Feature-engineered dataset saved!")
```
