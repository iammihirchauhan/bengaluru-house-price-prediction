```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/cleaned_bengaluru.csv")

df["price_per_sqft"] = (df["price"] * 100000) / df["total_sqft"]
df["sqft_per_bhk"] = df["total_sqft"] / df["bhk"]

print("Dataset shape:", df.shape)
```

```python
clean_df = df[
    (df["sqft_per_bhk"] >= 300) &
    (df["sqft_per_bhk"] <= 2000) &
    (df["price_per_sqft"] >= 500) &
    (df["price_per_sqft"] <= 50000) &
    (df["bath"] <= df["bhk"] + 2)
].copy()

print("Before filtering:", df.shape)
print("After filtering:", clean_df.shape)
print("Rows removed:", len(df) - len(clean_df))
print("Percentage removed:",
      round((len(df) - len(clean_df)) / len(df) * 100, 2), "%")
```

```python
print(clean_df.describe())

print("\nMissing values:")
print(clean_df.isnull().sum())

print("\nRemaining suspicious records:")
print(clean_df[
    (clean_df["sqft_per_bhk"] < 300) |
    (clean_df["price_per_sqft"] > 50000)
].shape)
```

```python
output_path = "../data/bengaluru_model_data.csv"

clean_df.to_csv(output_path, index=False)

print("Final dataset saved:", output_path)
```
